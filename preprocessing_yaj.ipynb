```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans,AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, adjusted_rand_score, normalized_mutual_info_score, calinski_harabasz_score,davies_bouldin_score

from sklearn.preprocessing import FunctionTransformer
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage

from sklearn.cluster import DBSCAN

from sklearn.neighbors import NearestNeighbors


import mlflow
import mlflow.sklearn

RANDOM_STATE = 42


sns.set(style="whitegrid")
```

```python
url = "Data/CC GENERAL.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.drop(columns=['CUST_ID'], inplace=True)
```

```python
df = df.dropna(subset=["CREDIT_LIMIT"])
```

```python
log_transform = FunctionTransformer(np.log1p, validate=False)

log_columns = [
    'BALANCE',
    'PURCHASES',
    'ONEOFF_PURCHASES',
    'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'MINIMUM_PAYMENTS',
    'CASH_ADVANCE_TRX',
    'PURCHASES_TRX'
]

normal_columns =[
    'BALANCE_FREQUENCY',
    'PURCHASES_FREQUENCY',
    'ONEOFF_PURCHASES_FREQUENCY',
    'PURCHASES_INSTALLMENTS_FREQUENCY',
    'CASH_ADVANCE_FREQUENCY',
    'PRC_FULL_PAYMENT',
    'TENURE'
]
```

```python
log_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("log", log_transform),
    ("scaler", StandardScaler())
])

normal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
processor = ColumnTransformer([
    ("log_features", log_pipeline, log_columns),

    ("normal_features", normal_pipeline, normal_columns )
])
```

```python
processor
```

```python
numeric_features = [
    'BALANCE',
    'BALANCE_FREQUENCY',
    'PURCHASES',
    'ONEOFF_PURCHASES',
    'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE',
    'PURCHASES_FREQUENCY',
    'ONEOFF_PURCHASES_FREQUENCY',
    'PURCHASES_INSTALLMENTS_FREQUENCY',
    'CASH_ADVANCE_FREQUENCY',
    'CASH_ADVANCE_TRX',
    'PURCHASES_TRX',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'MINIMUM_PAYMENTS',
    'PRC_FULL_PAYMENT',
    'TENURE'
]

preprocess_pipeline = Pipeline([
    ("preprocessor", processor)
])
```

```python
preprocess_pipeline
```

```python
X_scaled = preprocess_pipeline.fit_transform(df)
```

```python
pca = PCA(n_components=17, random_state=RANDOM_STATE)

X_pca = pca.fit_transform(X_scaled)
```

```python
cum_var = np.cumsum(pca.explained_variance_ratio_)
```

```python
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.plot(range(1, len(cum_var) + 1), cum_var, "o-", color="#5B8FB9")
ax.axhline(0.95, color="#E39482", linestyle="--", label="95% variance")
ax.set_xlabel("Number of components")
ax.set_ylabel("Cumulative explained variance")
ax.set_title("How many components actually capture the data?")
ax.legend()
plt.tight_layout()
plt.show()

n_comp_95 = int(np.argmax(cum_var >= 0.95) + 1)
print(f"Components needed for 95% variance: {n_comp_95} (out of {X_pca.shape[1]})")
```

```python
pca = PCA(n_components=10, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)
pca_2d = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca_2d = pca_2d.fit_transform(X_scaled)

# t-SNE visualization
tsne = TSNE(
    n_components=2,
    random_state=RANDOM_STATE,
    perplexity=30
)

X_tsne = tsne.fit_transform(X_scaled)
```

```python
mlflow.set_tracking_uri('sqlite:///mlflow_credit_card_clustering.db')
mlflow.set_experiment('Credit Card Clustering')
```

```python
K_range = range(2, 9)

inertias = []
sil_scores = []

for k in K_range:

    km = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    labels = km.fit_predict(X_pca)

    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_pca, labels))
```

```python
plt.figure(figsize=(6,4))
plt.plot(K_range, inertias, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("K-Means Elbow Method")
plt.grid(True)
plt.show()

plt.figure(figsize=(6,4))
plt.plot(K_range, sil_scores, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("K-Means Silhouette Scores")
plt.grid(True)
plt.show()
```

```python
with mlflow.start_run(run_name="KMeans"):

    kmeans_final = KMeans(
        n_clusters=2,
        random_state=RANDOM_STATE,
        n_init=10
    )

    kmeans_labels = kmeans_final.fit_predict(X_pca)

    mlflow.log_param("algorithm", "KMeans")
    mlflow.log_param("k", 2)
    mlflow.log_param("n_init", 10)
    mlflow.log_param("pca_components", 10)

    mlflow.log_metric(
        "silhouette_score",
        silhouette_score(X_pca, kmeans_labels)
    )

    mlflow.log_metric(
        "davies_bouldin_score",
        davies_bouldin_score(X_pca, kmeans_labels)
    )

    mlflow.sklearn.log_model(
        sk_model=kmeans_final,
        artifact_path="model"
    )

    # Elbow plot
    plt.figure(figsize=(6,4))
    plt.plot(K_range, inertias, marker="o")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Inertia")
    plt.title("K-Means Elbow Method")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("kmeans_elbow.png")
    mlflow.log_artifact("kmeans_elbow.png")
    plt.close()

    # Silhouette plot
    plt.figure(figsize=(6,4))
    plt.plot(K_range, sil_scores, marker="o")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Silhouette Score")
    plt.title("K-Means Silhouette Scores")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("kmeans_silhouette.png")
    mlflow.log_artifact("kmeans_silhouette.png")
    plt.close()

    # PCA
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_pca_2d[:,0],
        X_pca_2d[:,1],
        c=kmeans_labels,
        cmap="viridis",
        s=10
    )
    plt.title("K-Means Clusters (PCA)")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.tight_layout()
    plt.savefig("kmeans_pca_clusters.png")
    mlflow.log_artifact("kmeans_pca_clusters.png")
    plt.close()

    # t-SNE
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_tsne[:,0],
        X_tsne[:,1],
        c=kmeans_labels,
        cmap="viridis",
        s=10
    )
    plt.title("K-Means Clusters (t-SNE)")
    plt.xlabel("t-SNE Component 1")
    plt.ylabel("t-SNE Component 2")
    plt.tight_layout()
    plt.savefig("kmeans_tsne_clusters.png")
    mlflow.log_artifact("kmeans_tsne_clusters.png")
    plt.close()
```

```python
K_range = range(2, 9)

sil_scores = []

for k in K_range:

    agg = AgglomerativeClustering(
        n_clusters=k,
        linkage="ward"
    )

    labels = agg.fit_predict(X_pca)

    sil_scores.append(
        silhouette_score(X_pca, labels)
    )
```

```python
with mlflow.start_run(run_name="Hierarchical"):

    hierarchical_final = AgglomerativeClustering(
        n_clusters=2,
        linkage="ward"
    )

    hierarchical_labels = hierarchical_final.fit_predict(X_pca)

    mlflow.log_param("algorithm", "Hierarchical")
    mlflow.log_param("k", 2)
    mlflow.log_param("linkage", "ward")
    mlflow.log_param("pca_components", 10)

    mlflow.log_metric(
        "silhouette_score",
        silhouette_score(X_pca, hierarchical_labels)
    )

    mlflow.log_metric(
        "davies_bouldin_score",
        davies_bouldin_score(X_pca, hierarchical_labels)
    )

    mlflow.sklearn.log_model(
        sk_model=hierarchical_final,
        artifact_path="model"
    )

    # Silhouette plot
    plt.figure(figsize=(8,5))
    plt.plot(K_range, sil_scores, marker="o")
    plt.title("Hierarchical Clustering: Silhouette Score vs Number of Clusters")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Silhouette Score")
    plt.xticks(K_range)
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("hierarchical_silhouette.png")
    mlflow.log_artifact("hierarchical_silhouette.png")
    plt.close()

    # Dendrogram
    linked = linkage(X_pca, method="ward")

    plt.figure(figsize=(12,6))
    dendrogram(linked)
    plt.title("Hierarchical Clustering Dendrogram")
    plt.xlabel("Observations")
    plt.ylabel("Ward Distance")
    plt.tight_layout()
    plt.savefig("hierarchical_dendrogram.png")
    mlflow.log_artifact("hierarchical_dendrogram.png")
    plt.close()

    # PCA
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_pca_2d[:,0],
        X_pca_2d[:,1],
        c=hierarchical_labels,
        cmap="viridis",
        s=10
    )
    plt.title("Hierarchical Clusters (PCA)")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.tight_layout()
    plt.savefig("hierarchical_pca_clusters.png")
    mlflow.log_artifact("hierarchical_pca_clusters.png")
    plt.close()

    # t-SNE
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_tsne[:,0],
        X_tsne[:,1],
        c=hierarchical_labels,
        cmap="viridis",
        s=10
    )
    plt.title("Hierarchical Clusters (t-SNE)")
    plt.xlabel("t-SNE Component 1")
    plt.ylabel("t-SNE Component 2")
    plt.tight_layout()
    plt.savefig("hierarchical_tsne_clusters.png")
    mlflow.log_artifact("hierarchical_tsne_clusters.png")
    plt.close()
```

```python
nearest_neighbors = NearestNeighbors(n_neighbors=5)
nearest_neighbors.fit(X_pca)

distances, _ = nearest_neighbors.kneighbors(X_pca)
k_distances = np.sort(distances[:, -1])

plt.figure(figsize=(8, 5))
plt.plot(k_distances)
plt.title("DBSCAN k-Distance Plot")
plt.xlabel("Points sorted by distance")
plt.ylabel("5th Nearest Neighbour Distance")
plt.grid(True)
plt.show()
```

```python
eps_values = [1.5, 1.6, 1.7, 1.8, 1.9, 2.0]

sil_scores = []
n_clusters_list = []
noise_points = []

for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = dbscan.fit_predict(X_pca)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise = np.sum(labels == -1)

    n_clusters_list.append(n_clusters)
    noise_points.append(noise)

    if n_clusters > 1:

        silhouette = silhouette_score(
            X_pca[labels != -1],
            labels[labels != -1]
        )

    else:

        silhouette = np.nan

    sil_scores.append(silhouette)
```

```python
with mlflow.start_run(run_name="DBSCAN"):

    dbscan_final = DBSCAN(
        eps=2.0,
        min_samples=5
    )

    dbscan_labels = dbscan_final.fit_predict(X_pca)

    mlflow.log_param("algorithm", "DBSCAN")
    mlflow.log_param("k", 2)
    mlflow.log_param("eps", 2.0)
    mlflow.log_param("min_samples", 5)
    mlflow.log_param("pca_components", 10)

    mlflow.log_metric(
        "silhouette_score",
        silhouette_score(
            X_pca[dbscan_labels != -1],
            dbscan_labels[dbscan_labels != -1]
        )
    )

    mlflow.log_metric(
        "davies_bouldin_score",
        davies_bouldin_score(
            X_pca[dbscan_labels != -1],
            dbscan_labels[dbscan_labels != -1]
        )
    )

    mlflow.log_metric(
        "noise_points",
        np.sum(dbscan_labels == -1)
    )

    mlflow.sklearn.log_model(
        sk_model=dbscan_final,
        artifact_path="model",
    )

    # k-distance plot
    plt.figure(figsize=(8,5))
    plt.plot(k_distances)
    plt.title("DBSCAN k-Distance Plot")
    plt.xlabel("Points sorted by distance")
    plt.ylabel("5th Nearest Neighbour Distance")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("dbscan_k_distance.png")
    mlflow.log_artifact("dbscan_k_distance.png")
    plt.close()

    # Silhouette plot
    plt.figure(figsize=(8,5))
    plt.plot(eps_values, sil_scores, marker="o")
    plt.title("DBSCAN: Silhouette Score vs Epsilon")
    plt.xlabel("Epsilon")
    plt.ylabel("Silhouette Score")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("dbscan_silhouette.png")
    mlflow.log_artifact("dbscan_silhouette.png")
    plt.close()

    # PCA
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_pca_2d[:,0],
        X_pca_2d[:,1],
        c=dbscan_labels,
        cmap="viridis",
        s=10
    )
    plt.title("DBSCAN Clusters (PCA)")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.tight_layout()
    plt.savefig("dbscan_pca_clusters.png")
    mlflow.log_artifact("dbscan_pca_clusters.png")
    plt.close()

    # t-SNE
    plt.figure(figsize=(8,6))
    plt.scatter(
        X_tsne[:,0],
        X_tsne[:,1],
        c=dbscan_labels,
        cmap="viridis",
        s=10
    )
    plt.title("DBSCAN Clusters (t-SNE)")
    plt.xlabel("t-SNE Component 1")
    plt.ylabel("t-SNE Component 2")
    plt.tight_layout()
    plt.savefig("dbscan_tsne_clusters.png")
    mlflow.log_artifact("dbscan_tsne_clusters.png")
    plt.close()
mlflow.end_run()
```
